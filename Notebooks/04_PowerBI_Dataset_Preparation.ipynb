```python
import pandas as pd
```

```python
df = pd.read_csv("../Data/Processed/US_Accidents_Cleaned.csv")

df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
columns = [
    "Severity",
    "Start_Time",
    "Year",
    "Month_No",
    "Month",
    "Day",
    "Hour",
    "Weekday",
    "State",
    "City",
    "County",
    "Temperature(F)",
    "Humidity(%)",
    "Visibility(mi)",
    "Wind_Speed(mph)",
    "Weather_Condition",
    "Sunrise_Sunset",
    "Distance(mi)",
    "Amenity",
    "Bump",
    "Crossing",
    "Give_Way",
    "Junction",
    "No_Exit",
    "Railway",
    "Roundabout",
    "Station",
    "Stop",
    "Traffic_Calming",
    "Traffic_Signal",
    "Turning_Loop"
]

df = df[columns]
```

```python
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
numeric_cols = [
    "Temperature(F)",
    "Humidity(%)",
    "Visibility(mi)",
    "Wind_Speed(mph)"
]

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
categorical_cols = [
    "Weather_Condition",
    "Sunrise_Sunset"
]

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df["City"] = df["City"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df.to_csv(
    "../Data/Processed/PowerBI_Dataset.csv",
    index=False
)
```
