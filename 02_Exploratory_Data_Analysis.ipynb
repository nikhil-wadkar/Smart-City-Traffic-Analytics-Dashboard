```python
df["Hour"] = df["Start_Time"].dt.hour
```

```python
hourly = df["Hour"].value_counts().sort_index()

plt.figure(figsize=(12,5))

hourly.plot(kind="line", marker="o")

plt.title("Accidents by Hour")

plt.xlabel("Hour")

plt.ylabel("Accidents")

plt.grid(True)

plt.show()
```

```python
df["Day"] = df["Start_Time"].dt.day_name()
```

```python
day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

day_counts = df["Day"].value_counts().reindex(day_order)

plt.figure(figsize=(10,5))

day_counts.plot(kind="bar")

plt.title("Accidents by Day")

plt.show()
```

```python
df["Month"] = df["Start_Time"].dt.month_name()

month_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]

month_counts = df["Month"].value_counts().reindex(month_order)

plt.figure(figsize=(12,5))

month_counts.plot(kind="bar")

plt.title("Accidents by Month")

plt.show()
```

# Key Findings

- Severity 2 accidents dominate the dataset.
- California records the highest number of accidents.
- Most accidents occur during rush hours.
- Clear weather is the most common weather condition during accidents.
- Accident frequency varies across months and weekdays.
